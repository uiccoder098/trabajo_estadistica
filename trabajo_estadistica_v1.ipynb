### Descripción de las Variables del Dataset (Kaggle - Hotel Bookings)

A continuación se presenta la descripción detallada de cada una de las variables del conjunto de datos:

* **`hotel`**: Tipo de hotel (H1 = Resort Hotel o H2 = City Hotel).
* **`is_canceled`**: Valor que indica si la reserva fue cancelada (1) o no (0).
* **`lead_time`**: Número de días transcurridos entre la fecha de registro de la reserva en el sistema (PMS) y la fecha de llegada.
* **`arrival_date_year`**: Año de la fecha de llegada.
* **`arrival_date_month`**: Mes de la fecha de llegada.
* **`arrival_date_week_number`**: Número de semana del año para la fecha de llegada.
* **`arrival_date_day_of_month`**: Día del mes de la fecha de llegada.
* **`stays_in_weekend_nights`**: Número de noches de fin de semana (sábado o domingo) que el huésped se hospedó o reservó para hospedarse.
* **`stays_in_week_nights`**: Número de noches de entre semana (lunes a viernes) que el huésped se hospedó o reservó para hospedarse.
* **`adults`**: Número de adultos.
* **`children`**: Número de niños.
* **`babies`**: Número de bebés.
* **`meal`**: Tipo de régimen alimenticio reservado. Las categorías se presentan en paquetes estándar de hostelería:
  * `Undefined/SC`: Sin paquete de comidas (Solo alojamiento).
  * `BB` (Bed & Breakfast): Alojamiento y desayuno.
  * `HB` (Half board): Media pensión (desayuno y otra comida, habitualmente la cena).
  * `FB` (Full board): Pensión completa (desayuno, almuerzo y cena).
* **`country`**: País de origen. Las categorías se representan en formato ISO 3155–3:2013.
* **`market_segment`**: Designación del segmento de mercado. En las categorías, el término "TA" significa "Agentes de Viajes" (Travel Agents) y "TO" significa "Operadores Turísticos" (Tour Operators).
* **`distribution_channel`**: Canal de distribución de la reserva. El término "TA" significa "Agentes de Viajes" y "TO" significa "Operadores Turísticos".
* **`is_repeated_guest`**: Valor que indica si la reserva corresponde a un cliente repetido (1) o no (0).
* **`previous_cancellations`**: Número de reservas previas que fueron canceladas por el cliente antes de la reserva actual.
* **`previous_bookings_not_canceled`**: Número de reservas previas no canceladas por el cliente antes de la reserva actual.
* **`reserved_room_type`**: Código del tipo de habitación reservada. Se presenta el código en lugar de la designación por motivos de anonimato.
* **`assigned_room_type`**: Código del tipo de habitación asignada a la reserva. A veces el tipo de habitación asignada difiere de la reservada por razones operativas del hotel (ej. sobreventa) o por solicitud del cliente.
* **`booking_changes`**: Número de cambios o modificaciones realizadas a la reserva desde el momento en que se registró en el sistema hasta el momento del check-in o cancelación.
* **`deposit_type`**: Indicación de si el cliente realizó un depósito para garantizar la reserva. Esta variable puede tomar tres categorías:
  * `No Deposit`: No se realizó ningún depósito.
  * `Non Refund`: Se realizó un depósito por el valor total del costo de la estancia (no reembolsable).
  * `Refundable`: Se realizó un depósito por un valor inferior al costo total de la estancia.
* **`agent`**: ID de la agencia de viajes que realizó la reserva.
* **`company`**: ID de la empresa/entidad que realizó la reserva o responsable del pago de la misma. Se presenta el ID de forma anónima.
* **`days_in_waiting_list`**: Número de días que la reserva estuvo en lista de espera antes de ser confirmada al cliente.
* **`customer_type`**: Tipo de reserva, asumiendo una de las siguientes cuatro categorías:
  * `Contract`: Cuando la reserva tiene una tarifa de cupo (allotment) u otro tipo de contrato asociado.
  * `Group`: Cuando la reserva está asociada a un grupo.
  * `Transient`: Cuando la reserva no es parte de un grupo o contrato, y no está asociada a otra reserva transitoria.
  * `Transient-party`: Cuando la reserva es transitoria, pero está asociada al menos a otra reserva transitoria.
* **`adr`** (Average Daily Rate): Tarifa diaria promedio, calculada dividiendo la suma de todas las transacciones de alojamiento entre el número total de noches de estancia.
* **`required_car_parking_spaces`**: Número de plazas de aparcamiento requeridas por el cliente.
* **`total_of_special_requests`**: Número de solicitudes especiales realizadas por el cliente (ej. cama de matrimonio o piso alto).
* **`reservation_status`**: Último estado de la reserva, asumiendo una de las siguientes tres categorías:
  * `Canceled`: La reserva fue cancelada por el cliente.
  * `Check-Out`: El cliente realizó el ingreso, disfrutó de su estancia y ya se ha retirado.
  * `No-Show`: El cliente no se presentó (no-show) ni informó al hotel del motivo.
* **`reservation_status_date`**: Fecha en la que se estableció el último estado. Se puede usar junto con `reservation_status` para entender cuándo se canceló o se hizo el check-out.

## 2. Estadística descriptiva de variables

Descriptiva de 5 variables, analizando cada variable por separado.


**Variable 1: is_canceled (Cualitativa Binaria)**

```python
# ============================================================
# Variable 1: is_canceled (cualitativa binaria)
# Objetivo: resumir la variable con una tabla de frecuencias,
# medidas descriptivas y un gráfico de barras.
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Reutilizamos el dataframe que se carga al inicio del notebook
# (la celda que hace pd.read_csv define df_raw).
df = df_raw.copy()

# ------------------------------------------------------------
# 1. Tabla de frecuencias
# ------------------------------------------------------------
# Contamos cuántas reservas hay en cada categoría (0 y 1).
# reindex([0, 1]) garantiza que aparezcan las dos categorías
# aunque alguna tuviera cero casos.
n_valid = int(df['is_canceled'].notna().sum())
freq = df['is_canceled'].value_counts().reindex([0, 1], fill_value=0)

# Porcentaje sobre el total de valores válidos (no sobre nulos).
pct = (freq / n_valid * 100).round(2)

# Construimos una tabla legible para el lector.
tabla = pd.DataFrame({
    'Frecuencia absoluta': freq.values,
    'Porcentaje (%)': pct.values
}, index=['No cancelada (0)', 'Cancelada (1)'])

# Añadimos la fila de total y forzamos enteros en la frecuencia.
tabla.loc['Total'] = [int(freq.sum()), 100.00]
tabla['Frecuencia absoluta'] = tabla['Frecuencia absoluta'].astype(int)
display(tabla)

# ------------------------------------------------------------
# 2. Medidas descriptivas
# ------------------------------------------------------------
n = len(df)
n_missing = int(df['is_canceled'].isna().sum())
moda = int(df['is_canceled'].mode()[0])

# En una variable codificada como 0/1, la media es la proporción de unos.
proporcion = df['is_canceled'].mean()

print(f'n = {n:,} | valores perdidos = {n_missing}')
print(f'Moda = {moda} ({"cancelada" if moda == 1 else "no cancelada"})')
print(f'Proporción de cancelaciones = {proporcion:.4f} ({proporcion * 100:.2f}%)')

# ------------------------------------------------------------
# 3. Gráfico de barras
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7, 4))

# Una barra por categoría, con colores distintos para distinguirlas.
barras = ax.bar(
    ['No (0)', 'Sí (1)'],
    freq.values,
    color=['#4C72B0', '#C44E52']
)

# Escribimos el porcentaje encima de cada barra.
for barra, valor, porcentaje in zip(barras, freq.values, pct.values):
    ax.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + max(freq.values) * 0.01,
        f'{porcentaje:.1f}%',
        ha='center',
        va='bottom'
    )

# Títulos y etiquetas de los ejes.
ax.set_title('Distribución de is_canceled')
ax.set_xlabel('¿Reserva cancelada?')
ax.set_ylabel('Número de reservas')

# Quitamos el marco negro del lienzo y las marcas de los ejes.
for spine in ax.spines.values():
    spine.set_visible(False)
ax.tick_params(axis='both', length=0)

# Rejilla horizontal suave para facilitar la comparación de alturas.
ax.grid(axis='y', color='#E5E5E5', linewidth=0.8)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

El análisis de `is_canceled` muestra que **75.166 reservas (62,96 %)** no fueron canceladas frente a **44.224 (37,04 %)** que sí lo fueron. La categoría modal es “no cancelada” y no existen valores faltantes. La cancelación afecta a algo más de un tercio de las reservas, por lo que no es un fenómeno raro.

**Variable 4: lead_time (Cuantitativa Discreta)**

```python
# ============================================================
# Variable 4: lead_time (cuantitativa discreta)
# Días entre el registro de la reserva y la fecha de llegada.
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Reutilizamos el dataframe cargado al inicio del notebook.
df = df_raw.copy()

# Serie de trabajo: solo la variable que vamos a describir.
s = df['lead_time']

# ------------------------------------------------------------
# 1. Medidas descriptivas
# ------------------------------------------------------------
n = len(s)
n_valid = int(s.notna().sum())
n_missing = int(s.isna().sum())

media = s.mean()
mediana = s.median()
moda = s.mode().iloc[0]
desv = s.std()
minimo = s.min()
maximo = s.max()

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1
rango = maximo - minimo

# Coeficiente de variación: dispersión relativa respecto a la media.
cv = desv / media * 100

# Asimetría: si es positiva, la cola se alarga hacia la derecha.
asimetria = s.skew()

# Criterio de Tukey: outlier si está fuera de Q1 - 1,5·IQR o Q3 + 1,5·IQR.
lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr
outliers = s[(s < lim_inf) | (s > lim_sup)]

# Resumen en una tabla vertical para facilitar la lectura.
resumen = pd.Series({
    'n': n,
    'Valores perdidos': n_missing,
    'Media': media,
    'Mediana': mediana,
    'Moda': moda,
    'Desviación típica': desv,
    'Mínimo': minimo,
    'Q1': q1,
    'Q3': q3,
    'Máximo': maximo,
    'IQR': iqr,
    'Rango': rango,
    'Coef. variación (%)': cv,
    'Asimetría': asimetria,
    'Nº outliers (IQR)': len(outliers)
})
display(resumen.round(2).to_frame('Valor'))

print(f'Límite inferior (Tukey): {lim_inf:.1f} | Límite superior (Tukey): {lim_sup:.1f}')
print(f'Outliers: {len(outliers):,} ({len(outliers) / n_valid * 100:.2f}% de los válidos)')

print('\nPercentiles:')
print(s.quantile([0.5, 0.75, 0.9, 0.95, 0.99]).round(1).to_string())

print('\nValores más frecuentes:')
print(s.value_counts().head(10).to_string())

# ------------------------------------------------------------
# 2. Histograma
# ------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))

ax = axes[0]
ax.hist(s.dropna(), bins=50, color='#4C72B0', edgecolor='white')

# Líneas verticales para comparar media y mediana.
ax.axvline(media, color='#C44E52', linestyle='--', linewidth=1.5,
           label=f'Media = {media:.1f}')
ax.axvline(mediana, color='#55A868', linestyle='--', linewidth=1.5,
           label=f'Mediana = {mediana:.1f}')

ax.set_title('Histograma de lead_time')
ax.set_xlabel('Días de anticipación')
ax.set_ylabel('Número de reservas')
ax.legend(frameon=False)

# Quitar el marco negro y las marcas de los ejes.
for spine in ax.spines.values():
    spine.set_visible(False)
ax.tick_params(axis='both', length=0)
ax.grid(axis='y', color='#E5E5E5', linewidth=0.8)
ax.set_axisbelow(True)

# ------------------------------------------------------------
# 3. Diagrama de caja
# ------------------------------------------------------------
ax = axes[1]
ax.boxplot(
    s.dropna(),
    orientation='horizontal',
    patch_artist=True,
    widths=0.5,
    boxprops=dict(facecolor='#4C72B0', alpha=0.6, edgecolor='#4C72B0'),
    medianprops=dict(color='#C44E52', linewidth=2),
    whiskerprops=dict(color='#4C72B0'),
    capprops=dict(color='#4C72B0'),
    flierprops=dict(marker='o', markersize=3, alpha=0.35,
                    markerfacecolor='#C44E52', markeredgecolor='none')
)
ax.set_title('Diagrama de caja de lead_time')
ax.set_xlabel('Días de anticipación')
ax.set_yticks([])

# Quitar el marco negro y las marcas de los ejes.
for spine in ax.spines.values():
    spine.set_visible(False)
ax.tick_params(axis='both', length=0)
ax.grid(axis='x', color='#E5E5E5', linewidth=0.8)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

### Conclusión: lead_time (cuantitativa discreta)

El análisis de `lead_time` muestra una media de **104,01 días** y una mediana de **69 días**; la media es superior por la asimetría positiva (**1,35**). El 50 % central de las reservas se hizo entre **18 y 160 días** antes de la llegada (IQR = 142). El mínimo es **0 días** y el máximo **737**; el valor más frecuente es **0 días** (**6.345 reservas**), lo que indica una presencia importante de reservas de última hora. Se identificaron **3.005 valores atípicos altos (2,52 %)** por encima de **373 días**. La dispersión es muy elevada (desviación típica **106,86**; CV **102,74 %**).

**Variable 5: adr (Cuantitativa Continua)**

```python
# ============================================================
# Variable 5: adr (cuantitativa continua)
# ADR = Average Daily Rate: tarifa media diaria por noche.
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Reutilizamos el dataframe cargado al inicio del notebook.
df = df_raw.copy()

# Serie de trabajo: solo la variable que vamos a describir.
s = df['adr']

# ------------------------------------------------------------
# 1. Medidas descriptivas
# ------------------------------------------------------------
n = len(s)
n_valid = int(s.notna().sum())
n_missing = int(s.isna().sum())

media = s.mean()
mediana = s.median()
moda = s.mode().iloc[0]
desv = s.std()
minimo = s.min()
maximo = s.max()

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1
rango = maximo - minimo

# Coeficiente de variación: dispersión relativa respecto a la media.
cv = desv / media * 100

# Asimetría: positiva indica cola hacia tarifas altas.
asimetria = s.skew()

# Criterio de Tukey para valores atípicos.
lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr
outliers = s[(s < lim_inf) | (s > lim_sup)]

# Tabla resumen con formato legible (sin decimales innecesarios).
filas = [
    ('n', f'{n:,}'),
    ('Valores perdidos', f'{n_missing:,}'),
    ('Media', f'{media:,.2f}'),
    ('Mediana', f'{mediana:,.2f}'),
    ('Moda', f'{moda:,.2f}'),
    ('Desviación típica', f'{desv:,.2f}'),
    ('Mínimo', f'{minimo:,.2f}'),
    ('Q1', f'{q1:,.2f}'),
    ('Q3', f'{q3:,.2f}'),
    ('Máximo', f'{maximo:,.2f}'),
    ('IQR', f'{iqr:,.2f}'),
    ('Rango', f'{rango:,.2f}'),
    ('Coef. variación (%)', f'{cv:,.2f}'),
    ('Asimetría', f'{asimetria:,.2f}'),
    ('Nº outliers (IQR)', f'{len(outliers):,}'),
]
resumen = pd.DataFrame(filas, columns=['Medida', 'Valor']).set_index('Medida')
display(resumen)

print(f'Límite inferior (Tukey): {lim_inf:,.2f} | Límite superior (Tukey): {lim_sup:,.2f}')
print(f'Outliers: {len(outliers):,} ({len(outliers) / n_valid * 100:.2f}% de los válidos)')

print('\nPercentiles:')
print(s.quantile([0.5, 0.75, 0.9, 0.95, 0.99]).round(2).to_string())

print('\nValores más altos (top 10):')
print(s.sort_values(ascending=False).head(10).to_string())

print('\nCasos especiales:')
print(f'adr = 0: {int((s == 0).sum()):,} reservas')
print(f'adr < 0: {int((s < 0).sum()):,} reservas')
print(f'adr > 500: {int((s > 500).sum()):,} reservas')

# ------------------------------------------------------------
# 2. Histograma (0-500 € para poder ver la forma central)
# ------------------------------------------------------------
# Los valores negativos y los superiores a 500 se excluyen solo del
# histograma para que la escala sea legible; siguen en las medidas y
# en el diagrama de caja.
s_hist = s[(s >= 0) & (s <= 500)]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))

ax = axes[0]
ax.hist(s_hist, bins=50, color='#4C72B0', edgecolor='white')

# Líneas verticales para comparar media y mediana.
ax.axvline(media, color='#C44E52', linestyle='--', linewidth=1.5,
           label=f'Media = {media:.1f} €')
ax.axvline(mediana, color='#55A868', linestyle='--', linewidth=1.5,
           label=f'Mediana = {mediana:.1f} €')

ax.set_title('Histograma de adr (0-500 €)')
ax.set_xlabel('Tarifa media diaria (€)')
ax.set_ylabel('Número de reservas')
ax.legend(frameon=False)

# Quitar el marco negro y las marcas de los ejes.
for spine in ax.spines.values():
    spine.set_visible(False)
ax.tick_params(axis='both', length=0)
ax.grid(axis='y', color='#E5E5E5', linewidth=0.8)
ax.set_axisbelow(True)

# ------------------------------------------------------------
# 3. Diagrama de caja (con todos los datos)
# ------------------------------------------------------------
ax = axes[1]
ax.boxplot(
    s.dropna(),
    orientation='horizontal',
    patch_artist=True,
    widths=0.5,
    boxprops=dict(facecolor='#4C72B0', alpha=0.6, edgecolor='#4C72B0'),
    medianprops=dict(color='#C44E52', linewidth=2),
    whiskerprops=dict(color='#4C72B0'),
    capprops=dict(color='#4C72B0'),
    flierprops=dict(marker='o', markersize=3, alpha=0.35,
                    markerfacecolor='#C44E52', markeredgecolor='none')
)
ax.set_title('Diagrama de caja de adr')
ax.set_xlabel('Tarifa media diaria (€)')
ax.set_yticks([])

# Limitamos la vista para que la caja sea legible. Los 3 valores >500 €
# y el máximo de 5.400 € siguen en la tabla y en el texto de al lado.
ax.set_xlim(-20, 520)
ax.text(0.98, 0.90, '4 valores fuera de escala\n(máx. 5.400 €)',
        transform=ax.transAxes, ha='right', va='top',
        fontsize=9, color='#666666')

# Quitar el marco negro y las marcas de los ejes.
for spine in ax.spines.values():
    spine.set_visible(False)
ax.tick_params(axis='both', length=0)
ax.grid(axis='x', color='#E5E5E5', linewidth=0.8)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

print(f'\nHistograma: {len(s_hist):,} reservas ({len(s_hist) / n_valid * 100:.2f}%) entre 0 y 500 €.')
print(f'Fuera del histograma: {n_valid - len(s_hist):,} reservas ({(n_valid - len(s_hist)) / n_valid * 100:.2f}%) con adr < 0 o > 500 €.')
```

## 3. Relación entre dos variables


### Relación entre `lead_time` y `is_canceled`

**Pregunta:** ¿la anticipación con la que se reserva se relaciona con que la reserva se cancele?

Se comparan los estadísticos de `lead_time` en los dos grupos de `is_canceled` y se representan sus distribuciones. Este análisis es descriptivo: la prueba de significación se realiza en la Sección 4.


```python
# ============================================================
# Sección 3: relación entre lead_time y is_canceled
# ¿La anticipación de la reserva se relaciona con la cancelación?
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Reutilizamos el dataframe cargado al inicio del notebook.
df = df_raw.copy()

# Creamos una etiqueta legible para el estado de cancelación.
df['cancelacion'] = df['is_canceled'].map({0: 'No cancelada', 1: 'Cancelada'})

# ------------------------------------------------------------
# 1. Estadísticos de lead_time por grupo
# ------------------------------------------------------------
# Funciones auxiliares para los cuartiles.
def q25(x):
    return x.quantile(0.25)

def q75(x):
    return x.quantile(0.75)

# Calculamos las medidas para cada grupo de cancelación.
resumen = (
    df.groupby('cancelacion')['lead_time']
      .agg(
          n='count',
          media='mean',
          mediana='median',
          desv_tipica='std',
          minimo='min',
          q1=q25,
          q3=q75,
          maximo='max'
      )
)

# IQR y porcentaje sobre el total de reservas.
resumen['IQR'] = resumen['q3'] - resumen['q1']
resumen['% del total'] = (resumen['n'] / len(df) * 100)
resumen = resumen.reindex(['No cancelada', 'Cancelada'])

# Formateamos la tabla para que sea fácil de leer.
filas = []
for grupo in ['No cancelada', 'Cancelada']:
    r = resumen.loc[grupo]
    filas.append({
        'Grupo': grupo,
        'n': f'{int(r["n"]):,}',
        '% del total': f'{r["% del total"]:.2f} %',
        'Media': f'{r["media"]:,.2f}',
        'Mediana': f'{r["mediana"]:,.2f}',
        'Desv. típica': f'{r["desv_tipica"]:,.2f}',
        'Q1': f'{r["q1"]:,.2f}',
        'Q3': f'{r["q3"]:,.2f}',
        'IQR': f'{r["IQR"]:,.2f}',
        'Mínimo': f'{r["minimo"]:,.0f}',
        'Máximo': f'{r["maximo"]:,.0f}',
    })

tabla = pd.DataFrame(filas).set_index('Grupo')
display(tabla)

# ------------------------------------------------------------
# 2. Comparación directa entre grupos
# ------------------------------------------------------------
mediana_no = resumen.loc['No cancelada', 'mediana']
mediana_si = resumen.loc['Cancelada', 'mediana']

print(f'Diferencia de medianas (cancelada - no cancelada): {mediana_si - mediana_no:,.2f} días')
print(f'Cociente de medianas (cancelada / no cancelada): {mediana_si / mediana_no:,.2f}')

# Porcentaje de reservas con más de 100 días de anticipación en cada grupo.
for grupo in ['No cancelada', 'Cancelada']:
    sub = df.loc[df['cancelacion'] == grupo, 'lead_time']
    porcentaje = (sub > 100).mean() * 100
    print(f'{grupo}: {porcentaje:.2f} % con lead_time > 100 días')

# ------------------------------------------------------------
# 3. Gráfico 1: diagrama de caja por grupo
# ------------------------------------------------------------
grupo_no = df.loc[df['cancelacion'] == 'No cancelada', 'lead_time'].dropna()
grupo_si = df.loc[df['cancelacion'] == 'Cancelada', 'lead_time'].dropna()

fig, axes = plt.subplots(1, 2, figsize=(12, 4.4))

ax = axes[0]
bp = ax.boxplot(
    [grupo_no, grupo_si],
    positions=[1, 2],
    patch_artist=True,
    widths=0.5,
    boxprops=dict(edgecolor='#4C72B0'),
    medianprops=dict(color='#C44E52', linewidth=2),
    whiskerprops=dict(color='#4C72B0'),
    capprops=dict(color='#4C72B0'),
    flierprops=dict(marker='o', markersize=2, alpha=0.2,
                    markerfacecolor='#C44E52', markeredgecolor='none')
)

# Coloreamos cada caja y anotamos su mediana.
colores = ['#4C72B0', '#C44E52']
for caja, color in zip(bp['boxes'], colores):
    caja.set_facecolor(color)
    caja.set_alpha(0.55)

for posicion, datos in zip([1, 2], [grupo_no, grupo_si]):
    ax.text(posicion, datos.median() + 5, f'{datos.median():.0f}',
            ha='center', va='bottom', fontsize=9)

ax.set_xticks([1, 2])
ax.set_xticklabels(['No cancelada', 'Cancelada'])
ax.set_ylim(0, 365)
ax.set_title('lead_time según cancelación')
ax.set_ylabel('Días de anticipación')

# ------------------------------------------------------------
# 4. Gráfico 2: distribuciones superpuestas (0-365 días)
# ------------------------------------------------------------
ax = axes[1]
bins = range(0, 370, 10)

# density=True permite comparar las formas aunque los grupos tengan distinto n.
ax.hist(grupo_no, bins=bins, density=True, alpha=0.55,
        color='#4C72B0', edgecolor='white', label='No cancelada')
ax.hist(grupo_si, bins=bins, density=True, alpha=0.55,
        color='#C44E52', edgecolor='white', label='Cancelada')

# Líneas verticales con las medianas de cada grupo.
ax.axvline(grupo_no.median(), color='#4C72B0', linestyle='--', linewidth=1.5)
ax.axvline(grupo_si.median(), color='#C44E52', linestyle='--', linewidth=1.5)

ax.set_xlim(0, 365)
ax.set_title('Distribución de lead_time por grupo')
ax.set_xlabel('Días de anticipación')
ax.set_ylabel('Densidad')
ax.legend(frameon=False)

# ------------------------------------------------------------
# 5. Estilo limpio para los dos paneles
# ------------------------------------------------------------
for ax in axes:
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.tick_params(axis='both', length=0)
    ax.grid(axis='y', color='#E5E5E5', linewidth=0.8)
    ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

# Aviso sobre los valores que quedan fuera de la vista de los gráficos.
n_fuera = int((df['lead_time'] > 365).sum())
print(f'Nota: los gráficos muestran 0-365 días; {n_fuera:,} reservas '
      f'({n_fuera / len(df) * 100:.2f} %) superan los 365 días y se excluyen solo de la visualización.')
```

## 4. Contraste de hipótesis o ACP


### Contraste de hipótesis: Mann-Whitney U

**Pregunta:** ¿la diferencia de `lead_time` entre reservas canceladas y no canceladas es estadísticamente significativa?

- **H0:** la distribución de `lead_time` es la misma en ambos grupos.
- **H1:** las distribuciones difieren.
- **Prueba:** Mann-Whitney U, dos colas.
- **Nivel de significación:** α = 0,05.

Se elige una prueba no paramétrica porque `lead_time` es muy asimétrica y no sigue una distribución normal.


```python
# ============================================================
# 4.1 Contraste de hipótesis: Mann-Whitney U
# H0: la distribución de lead_time es la misma en ambos grupos.
# H1: las distribuciones difieren.
# ============================================================

import numpy as np
from scipy.stats import mannwhitneyu

# Separamos lead_time en los dos grupos de cancelación.
no_cancelada = df.loc[df['is_canceled'] == 0, 'lead_time'].dropna()
cancelada = df.loc[df['is_canceled'] == 1, 'lead_time'].dropna()

# Contraste de dos colas.
resultado = mannwhitneyu(cancelada, no_cancelada, alternative='two-sided')

U = resultado.statistic
p = resultado.pvalue

# Tamaño del efecto: delta de Cliff (equivalente a la correlación rango-biserial).
n1 = len(cancelada)
n2 = len(no_cancelada)
delta = 2 * U / (n1 * n2) - 1

print(f'n canceladas = {n1:,}')
print(f'n no canceladas = {n2:,}')
print(f'U = {U:,.1f}')
print(f'p-valor = {p:.3e}' if p > 0 else 'p-valor < 1e-300')
print(f'Delta de Cliff = {delta:.3f}')

alpha = 0.05
if p < alpha:
    print(f'Con alpha = {alpha}, se rechaza H0: hay evidencia de diferencia entre las distribuciones.')
else:
    print(f'Con alpha = {alpha}, no se rechaza H0.')

# Comparación descriptiva que acompaña al contraste.
print(f'Mediana canceladas = {cancelada.median():.2f} días')
print(f'Mediana no canceladas = {no_cancelada.median():.2f} días')
print(f'Diferencia de medianas = {cancelada.median() - no_cancelada.median():.2f} días')

# Interpretación orientativa del tamaño del efecto (Romano et al.).
if abs(delta) < 0.147:
    magnitud = 'despreciable'
elif abs(delta) < 0.33:
    magnitud = 'pequeño'
elif abs(delta) < 0.474:
    magnitud = 'mediano'
else:
    magnitud = 'grande'
print(f'Magnitud del efecto (delta de Cliff): {magnitud}')
```

### Interpretación del contraste

El contraste de Mann-Whitney U indica que las distribuciones de `lead_time` difieren de forma estadísticamente significativa entre reservas canceladas y no canceladas (**U = 2.291.186.266,5; p < 0,001**; α = 0,05). Por tanto, se rechaza H0.

La mediana de las canceladas es **113 días**, frente a **45 días** en las no canceladas: una diferencia de **68 días**. El tamaño del efecto (**delta de Cliff = 0,379**) es **mediano**, lo que indica que la diferencia no solo es significativa, sino también relevante en magnitud.

Con una muestra tan grande, el p-valor es extremadamente pequeño; por eso la interpretación debe apoyarse en la mediana, el IQR y el tamaño del efecto, no solo en la significación.
